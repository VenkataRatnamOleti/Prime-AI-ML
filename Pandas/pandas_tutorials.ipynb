```python
# selecting date

#columns
df["city"]
df[["city","aqi"]]  # columns names are case sensitive


#rows
df.loc[2]
df.loc[0:2]  # start idx : end idx [inclusive]


df.iloc[2]
df.iloc[0:3] # end idx exclusive


df.loc[0:2,["city","aqi"]]
df.iloc[0:3,[1,2,3,4]] #position
df.iloc[0:3,1:5]


# single scalar values (at & iat)

df.at[0,"city"]
df.iat[0,2]


df["city"]   # view, not copy

cities = df["city"]
cities[0] = "New York"
df


# copy

cities = df["city"].copy()
cities[0] = "Mumbai"

df
cities
```

```python
# Filtering of Data

df[ (df["aqi"] > 100) & (df["temperature"] > 30) ]
df[ (df["aqi"] > 100) & (df["temperature"] < 30) ]


aqi_data = df[ (df["aqi"] > 100) & (df["temperature"] > 30) ][["city","aqi"]]

aqi_data
aqi_data.iloc[0]
aqi_data.loc[6]
```

```python
# filtering data using query

# Query
df.query("aqi>100")

# Rules
# 1. Expressed as a string
# 2. Column based reference
# 3. Backticks for col_name with space or special chars
# 4. Operators (&, |, ~, >=, >, <=, <, == etc.)
# 5. @ to reference Python vars
# 6. Chained comparisons

df.query("aqi>100 & temperature > 30")[["city","aqi"]]


aqi_val = 100
df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]]


# query returns copy. not view.
```
