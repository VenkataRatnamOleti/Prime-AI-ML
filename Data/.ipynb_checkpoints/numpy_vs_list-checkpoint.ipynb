```python
import numpy as np
import time
import sys
```

```python
# Performance of List VS NP Array (Time)

size = 10_000_000
py_list = list(range(size))

np_arr = np.array(py_list)


# Squaring Operaions
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()

print(f"Time for Squaring the Python List = {end-start} Seconds")

start = time.time()
sq_array = np_arr ** 2
end = time.time()

print(f"Time for Squaring the NumPy Array = {end-start} Seconds")

```

```python
# Memory Performance
print(f"Size of Python List {sys.getsizeof(py_list)*len(py_list)} bytes") #Extra size for metadata
print(f"Size of NumPy Array {np_arr.size} bytes")
```
