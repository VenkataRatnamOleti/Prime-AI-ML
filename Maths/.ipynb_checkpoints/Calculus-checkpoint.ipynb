```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,100,100_000)
y = np.sqrt(x)
```

```python
plt.plot(x,y,label="y = ✓x")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Y = ✓X")
plt.legend()
plt.grid(True)

plt.show()
```

```python
x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = 2*np.sin(x)
x1 = np.sin(x)
x2 = -2*np.sin(x)
x3 = np.sin(x)+2
```

```python
plt.plot(x,y,label="2sin(x)",color="blue")
plt.plot(x,x1,label="sin(x)",color="red")
plt.plot(x,x2,label="-2sin(x)",color="purple")
plt.plot(x,x3,label="sin(x)+2",color="green")


plt.axvline(0,color="black",linewidth=1.5,linestyle="--",label="Y - axis")
plt.axhline(0,color="black",linewidth=1.5,linestyle="--",label="X - axis")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = sin(x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
x = np.linspace(0,5,)
y = np.square(x)
y2 = np.square(x+2)
```

```python
plt.plot(x,y,label="x*x",color="blue")
plt.plot(x,y2,label="x*x + 2",color="blue")

plt.axvline(0,color="black",linewidth=1.5,linestyle="--",label="Y - axis")
plt.axhline(0,color="black",linewidth=1.5,linestyle="--",label="X - axis")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Graph of y = x*x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
x = np.linspace(-5,10,1000)
y = np.square(x)-(4*x)+5
# y2 = np.square(x+2)

plt.plot(x,y,color="blue")


plt.axvline(0,color="black",linewidth=1.5,linestyle="--",label="Y - axis")
plt.axhline(0,color="black",linewidth=1.5,linestyle="--",label="X - axis")
```
