```python
from scipy.stats import binom
```

```python
n = 5
k = 3
p = 0.5
```

```python
prob = binom.pmf(k,n,p)   # -> Binomial Distribution
```

```python
print(prob)
```

```python
# Uniform Distribution

import numpy as np
import matplotlib.pyplot as plt
```

```python
values = np.random.uniform(0,10,10_00_000)

plt.hist(values,bins=100, density=True,alpha=0.3)

plt.title("Continuous Uniform Distribution between 0 and 10")
plt.xlabel("x")
plt.ylabel("Probabiity Density")
plt.grid(True)

plt.show()
```
