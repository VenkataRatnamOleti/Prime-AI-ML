```python
import numpy as np
```

```python
# Linear Regression

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter= n_iter

    def fit(self, X, Y):
        m, n = X.shape

        # step 1
        self.bias = 0
        self.weights = np.zeros(n)

        # Gradient Descent
        for i in range(self.n_iter):
            # step 2 - cac y_pred
            y_pred = self.bias + np.dot(X, self.weights)
    
            # step 3 - calc gradient
            db = (1/m)*np.sum(y_pred-Y)
            dw = (1/m)*np.dot(X.T,(y_pred-Y))
    
            # step 4 - convergence theorem + params update
            self.bias -= self.lr*db
            self.weights -= self.lr*dw


    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1],[2],[3],[4],[5]])
Y = np.array([2,4,6,8,10])

model = LinearRegression()
model.fit(X,Y)

y_pred = model.predict(X)
print(y_pred)
```

```python
print(model.bias)
print(model.weights)
```

## Linear Regression
### Another way is OLS (Ordinary Least Squares). OLS is replaced with Gradient Descent.

- It works with small data

# OLS -> Normal Equation

### 0 = (X.T * X)^(-1) * (X.T * Y)

```python
# Linear Regression with OLS

class LinearRegressionOLS:
    def __init__(self):
        self.bias = None
        self.weights = None

    def fit(self, X, Y):
        m, n = X.shape
        X_b = np.c_[np.ones((m,1)),X]

        # Normal Equation
        # theta = np.dot(np.linalg.inv(np.dot(X_b.T,X_b)),np.dot(X_b.T,Y))
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ Y    # -> Time Compltexity is very large and exprensive [O(n^3)]

        # time complexity of GD is [O(n*m)]
        self.bias = theta[0]
        self.weights = theta[1:]
        
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights)
        return y_pred
```

```python
X = np.array([[1],[2],[3],[4],[5]])
Y = np.array([2,4,6,8,10])

model = LinearRegressionOLS()
model.fit(X,Y)

y_pred = model.predict(X)
print(y_pred)
```

```python
print(model.bias)
print(model.weights)
```

# Gradient Descent VS OLS

### When is LR with GD better than OLS ?
    - When dataset is very large (Big Data)
    - When fetures are many (High Dimensional)
    - When Regularization is required
    - When data is streaming / online learning
    - When dataset does NOT fit in memory
