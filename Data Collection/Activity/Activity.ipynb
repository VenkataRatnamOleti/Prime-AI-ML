```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    # print(res)
    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div.quote")
    
    if not quotes:
        print("No valid pages any more..")
        break

    with open(f"quotes{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"downloaded data from page {page_count}")

    page_count  += 1
```

```python
# Extract useful info
n = page_count
life_quotes = []

for i in range(1,n):
    with open(f"quotes{i}.html","r",encoding="utf-8") as f:
        html_content = f.read()
    
    soup = BeautifulSoup(html_content,"lxml") 
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = []
    
        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())
    
        # print(all_tags)
    
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text,author])
```
