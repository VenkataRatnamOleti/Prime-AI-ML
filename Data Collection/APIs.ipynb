```python
import requests

URL = "https://dattebayo-api.onrender.com/tailed-beasts"

res = requests.get(URL)
```

```python
res
# 200 is status code
# 404 page not found
# 500 internal server error
json_data = res.json()
```

```python
import pandas as pd

df = pd.json_normalize(json_data["tailed-beasts"])
```

```python
df.columns
df = df[['id','name','jutsu','natureType']]
```
