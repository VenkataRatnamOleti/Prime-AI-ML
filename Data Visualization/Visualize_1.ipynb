### Pie Charts
A pie chart is a circular chart divided into slices to show relative proportions of a whole.
- Each slice = a category
- Size of slice = value of that category relative to total

  Prefer them only for few categories.
  Max 5-6 categories.

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [540, 150, 120, 100, 50]

plt.style.use("dark_background")
plt.pie(amounts, labels=expenses)
```

## Pie Charts Customizations
- autopct = "%1.1f%%" -> shows percentage of each slice
- colors -> add custom colours
- wedgeprops -> define (edgecolor, linewidth, linestyle, alpha, fill...)
- explode -> highlights a slice by pulling it out
- startangle -> rotate the chart
- shadow -> adds 3D effect or makes the chart circular

```python
plt.style.use("default")
explode = [0, 0.4, 0, 0.2, 0]
colors = ["red","magenta", "pink", "blue", "yellow"]
plt.pie(amounts, labels=expenses, autopct="%1.1f%%", colors=colors, wedgeprops={
    "edgecolor":"purple",
    "linewidth":2,
    "linestyle":"--",
    "alpha":0.5,
    "fill":True
}, explode = explode, startangle=180, shadow=True)

# wedge -> each slice
```

### Histograms
A histogram shows the distribution of numerical data by:
- Dividing Values into bins (ranges)
- Counting how many values fall into each bin (frequency)

```python
# Histogram

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

scores = np.random.normal(70,10,100) # normal distribution
```

```python
# plt.hist(scores,color="skyblue",edgecolor="black",bins=20)
bins = [0,30,70,90,100]
plt.hist(scores,color="skyblue",edgecolor="black",bins=bins)
plt.ylabel("No. of Students")
plt.xlabel("Scores of Students")
```

```python
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 52, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.hist(marks,color="skyblue",edgecolor="black",bins=bins)

plt.axvline(
    x=33,
    label="Passing Marks (33)",
    color="red",
    linewidth=2,
    linestyle="--"
)
plt.ylabel("No. of Students")
plt.xlabel("Marks of Students")
plt.title("Distribution of Marks and Students")
```

```python
# Multiple dataset in Histograms
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99
]

fraud_transactions = [
    50, 100, 150, 200, 300
]

plt.hist(legit_transactions,color="green",bins=20,label="legit", edgecolor="black",alpha=0.5)
plt.hist(fraud_transactions,color="red",bins=20,label="fraud", edgecolor="black",alpha=0.5)
plt.legend()

# Mistake to be avoided
# 1. bin value should not be too small and too large.
# 2. Histograms are used for numerical data not categories.
```

### Box Plots
a Box plot is a statistical visualization that summarizes the distribution of a dataset using five key numbers:
- Minimum (lowest non-outlier)
- first quartile (Q1) - 25th percentile
- Median (Q2) - 50th percentile
- third quartile (Q3) - 75th percentile
- maximum (Highest non-outlier)

```python
# Box plot (or) Whiskers plot

data = [7, 8, 5, 6, 9, 4, 10, 12, 15]
plt.boxplot(data)
plt.grid()


# whiskers
# lower whisker = Q1 - (1.5 * IQR)
# Upper whisker = Q3 + (1.5 * IQR)
# IQR -> Inter Quartile Range = Q3 - Q1;




# outliers
```

### Operations on Box Plot
1. Multiple Datasets
2. Horizontal Box Plot. -> Default : 1.5
3. Show Mean
4. Change Whisker Length

```python
# Multiple datasets
group1 = np.random.normal(50,10,100)
group2 = np.random.normal(60,15,100)

plt.boxplot([group1,group2],tick_labels=["group1","group2"])
plt.grid(True)
plt.show()
plt.boxplot(data,vert=False,showmeans=True, whis=2.0)
```

### Stack Plots
    A stack plot is a type of plot where multiple data series are stacked on top of each other.
    each "layer" shows the contribution of one category, and the top line shows the cumulative total.

```python
days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
direct = [50,60,70,80,90,100,110]
organic = [30,40,50,55,60,70,80]
social = [20,25,30,35,40,50,60]


plt.stackplot(days,direct,organic,social,labels=["Direct","Organic","Social"])
plt.title("Marketing data for the week")
plt.xlabel("Days")
plt.ylabel("No. of customers")
plt.legend()
plt.grid()
```

### Subplots
    A subplot allows you to draw multiple plots in the same figure.

            plt.subplot(nrows,ncols,index)

```python
# subplots

x = [i for i in range(1,10000000)]
y1 = [np.sqrt(i) for i in x] #square root
y2 = [i*2 for i in x] #double
y3 = [i**2 for i in x] # square
y4 = [i**3 for i in x] # cubes

plt.subplot(2,2,1)
plt.plot(x,y1)
plt.title("plot1 - square root")

plt.subplot(2,2,2)
plt.plot(x,y2)
plt.title("plot12 - double")

plt.subplot(2,2,3)
plt.plot(x,y3)
plt.title("plot3 - squares")

plt.subplot(2,2,4)
plt.plot(x,y4)
plt.title("plot4 - cubes")

plt.tight_layout()
```

## Modern Matplotlib
#### (Object Oriented API)

        plt. -> state based API

        plt.subplots()     ---> returns figure and axes

```python
x = [i for i in range(1,11)]
y1 = [np.sqrt(i) for i in x] #square root
y2 = [i*2 for i in x] #double
y3 = [i**2 for i in x] # square
y4 = [i**3 for i in x] # cubes

plt.subplot(2,2,1)
plt.plot(x,y1)
plt.title("plot1 - square root")

plt.subplot(2,2,2)
plt.plot(x,y2)
plt.title("plot12 - double")

plt.subplot(2,2,3)
plt.plot(x,y3)
plt.title("plot3 - squares")

plt.subplot(2,2,4)
plt.plot(x,y4)
plt.title("plot4 - cubes")

plt.tight_layout()
```

```python
fig, ax = plt.subplots(2,2,figsize=[7,7])

ax[0][0].plot(x,y1)
ax[0][0].set_title("Plot1 - Square root")

ax[0][1].plot(x,y2)
ax[0][1].set_title("Plot2 - Double")

ax[1][0].plot(x,y3)
ax[1][0].set_title("Plot3 - Square")

ax[1][1].plot(x,y4)
ax[1][1].set_title("Plot4 - Cube")

fig.tight_layout
fig.suptitle("Multiple Plots")
```

## Activity
    compare temperature of cities over a week using subplots & loops.

```python
days = ["Mon","Tue","Wed","Thu","Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22,23,21,24,25],
    [18,19,17,20,21],
    [30,32,31,33,34],
    [25,26,24,27,28]
]
```

```python
fig, ax = plt.subplots(2,2,figsize=[5,5])
k = 0
for i in range(2):
    for j in range(2):
        ax[i][j].plot(days,temperatures[k],marker='o')
        ax[i][j].set_title(f"{cities[k]}")
        ax[i][j].grid()
        k += 1

fig.tight_layout()
fig.suptitle("Temperature in cities over the week")
```

```python
fig, axes = plt.subplots(2,2)

axes = axes.flatten()

for i,ax in enumerate(axes): # index and value at index returns
    ax.plot(days,temperatures[i],marker='o');
    ax.set_title(cities[i])
    ax.grid()

fig.tight_layout()
```
