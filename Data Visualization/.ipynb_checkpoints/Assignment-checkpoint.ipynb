```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.random.normal(loc=70,scale=10, size=200)
mean = np.mean(data)
plt.hist(data,bins=30,edgecolor="black",color="skyblue")
plt.axvline(x=mean,color="red",linestyle="dashed")
```

```python
import seaborn as sns
penguins = sns.load_dataset("penguins")
```

```python
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species"
)
```

```python
tips = sns.load_dataset("tips")
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

# sunday has the highest median bill
# there are more outliers on thursday
```

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker"
)

plt.title("Total Bill Vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
```

```python
pivot_table = tips.pivot_table(index="day",columns="time",values="tip",aggfunc="mean")
sns.heatmap(
    pivot_table,
    annot=True
    # cmap="coolwarm"
)
```

```python
# the sunday dinner day-time combination performs the best.
```
