### Other Important Plot Methods
- plt.plot(x,y) - Line plot
- plt.bar(x,height) - Vertical bar chart
- plt.barh(y,width) - Horizontal bar chart
- plt.scatter(x,y) - Scatter plot
- plt.pie(data) - Pie chart
- plt.hist(data,bins) - Histogram
- plt.boxplot(data) - Box plot

```python
# Bar Charts

# A bar plot( or bar chart ) is a graph that displays categorical data using rectangular bars.
# Each bar's height (or length, in horizontal bars) represents a value.
```

```python
movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M
```

```python
plt.bar(years,revenue,color="green")
plt.title("Revenue in Each Year for Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue in $M")
for i in range(len(years)):
    plt.text(years[i], revenue[i]+5, revenue[i],ha="center") # bar lables (x-axis pos, y-axiz pos,text, alignment)


plt.ylim(0,max(revenue)+100)
```

```python
# Compare 2 Sets of Data

# We customise width. width controls how tick each bar is. Usually ranges from 0 to 1.
# [Default val is 0.8]
```

```python
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378,2788,829,185,275] # in $M

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M
```

```python
import numpy as np
width = 0.25
x = np.arange(len(years))

# for i in range(len(years))

plt.bar(x-width,revenue,width,label="Oscar Movies")  # shifting to left
plt.bar(x,non_oscar_revenue,width,label="Non-Oscar Movies") # shifting to right
plt.bar(x+width,revenue,width,label="Oscar Movies")

plt.title("Oscar vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue($M)")
plt.legend()
```

```python
# Horizontal Bar Chart

plt.barh(movies,revenue)
plt.xlabel("Revenue")
plt.ylabel("Movie Names")
```

```python
# Scatter Plots
# A scatter plot shows the relationship between two variables by plotting points on a 2D plane

# corelation
# outliers
# clusters
```

```python
people = ["Person A","Person B","Person C","Person D","Person E",
          "Person F","Person G","Person H","Person I","Person J"]
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]
```

```python
plt.scatter(age,bp)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

### Pie Charts
A pie chart is a circular chart divided into slices to show relative proportions of a whole.
- Each slice = a category
- Size of slice = value of that category relative to total

  Prefer them only for few categories.
  Max 5-6 categories.

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [540, 150, 120, 100, 50]

plt.style.use("dark_background")
plt.pie(amounts, labels=expenses)
```

## Pie Charts Customizations
- autopct = "%1.1f%%" -> shows percentage of each slice
- colors -> add custom colours
- wedgeprops -> define (edgecolor, linewidth, linestyle, alpha, fill...)
- explode -> highlights a slice by pulling it out
- startangle -> rotate the chart
- shadow -> adds 3D effect or makes the chart circular

```python
plt.style.use("default")
explode = [0, 0.4, 0, 0.2, 0]
colors = ["red","magenta", "pink", "blue", "yellow"]
plt.pie(amounts, labels=expenses, autopct="%1.1f%%", colors=colors, wedgeprops={
    "edgecolor":"purple",
    "linewidth":2,
    "linestyle":"--",
    "alpha":0.5,
    "fill":True
}, explode = explode, startangle=180, shadow=True)

# wedge -> each slice
```
