# SEABORN

## Seaborn is a Python data visualization library based on matplotlib.


- matplotlib is used for low level work
- seaborn is used for dealing high level work

## Benefits of Seaborn
1. In-built Datasets are available
2. High level features
3. great support for pandas dataframes
4. boxpots, heapmaps

### pip install seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
# sns.get_dataset_names()
tips = sns.load_dataset("tips")
type(tips)
print(tips.head())
```

```python
# relation plot (relplot()) -> scatter plot, line plot etc,.
# sns.set_theme(style="ticks")

sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
x_vals = [i for i in range(1,11)]
x_sq = [i**2 for i in x_vals]

sns.set_theme(style="ticks")
sns.relplot(
    x=x_vals,
    y=x_sq,
    kind="line",
)
```

```python
## Relational plots in seaborn

# 1. Scatter Plots
# 2. Linr Plots
```

```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
# line plot

flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)

# Shaded area
# confidence interval (ci) -> 95%
# mean + or - ci
```

```python
# Categorical Plots
# 1. Bar Plots
# 2. Box Plots
```

```python
# print(tips.head())

sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
# Box Plot

sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
# Distribution plots

# 1. Histogram
```

```python
penguins = sns.load_dataset("penguins")
# penguins.head()

sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

# Matrix Plots
- Heatmaps - Plot rectangular data as color-encoded matrix

```python
flights.head()
```

```python
flights_pivot = flights.pivot(index="month",columns="year",values="passengers")
```

```python
sns.heatmap(
    flights_pivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)

plt.title("Passenders Heatmap")
plt.tight_layout()
```

# Best Practices of using Seaborn

```python
# Seaborn is used for individual plot
# but if need any low level customizations then we have to use matplotlib
```

```python
fig, ax = plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    ax=ax,
    ci=None
)

ax.set_title("Avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")

fig.tight_layout()
```
